```python
import numpy as np
import pandas as pd
import re
import nltk
from nltk.tokenize import RegexpTokenizer
from nltk.corpus import stopwords
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report

# Ensure necessary NLTK downloads
nltk.download('stopwords')

# Load dataset
file_path = "/content/sample_data/train.csv"
df = pd.read_csv(file_path, encoding='latin1')

# Initialize tokenizer
tokenizer = RegexpTokenizer(r'\w+')

# Text Preprocessing Function
def preprocess_text(text):
    text = str(text).lower()  # Convert to lowercase
    text = re.sub(r'http\S+', '', text)  # Remove URLs
    text = re.sub(r'@\w+', '', text)  # Remove @mentions
    text = re.sub(r'[^a-zA-Z\s]', '', text)  # Remove special characters
    tokens = tokenizer.tokenize(text)  # Tokenization
    tokens = [word for word in tokens if word not in stopwords.words('english')]  # Remove stopwords
    return " ".join(tokens)

# Apply text preprocessing
df['clean_text'] = df['SentimentText'].astype(str).apply(preprocess_text)

# Define features and labels
X = df['clean_text']
y = df['Sentiment']

# Convert text to numerical representation
vectorizer = TfidfVectorizer()
X_vectorized = vectorizer.fit_transform(X)

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(X_vectorized, y, test_size=0.2, random_state=42)

# Train Softmax Classifier (Logistic Regression for multiclass classification)
model = LogisticRegression(multi_class='multinomial', solver='lbfgs', max_iter=1000)
model.fit(X_train, y_train)

# Make Predictions
y_pred = model.predict(X_test)

# Compute Standard Metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred, average='weighted')
recall = recall_score(y_test, y_pred, average='weighted')  # Sensitivity
f1 = f1_score(y_test, y_pred, average='weighted')

# Compute Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:\n", cm)

# Compute Additional Metrics
num_classes = cm.shape[0]  # Number of classes

specificities = []
npvs = []

for i in range(num_classes):
    # True Positives, False Positives, False Negatives, True Negatives for each class
    tp = cm[i, i]
    fp = cm[:, i].sum() - tp
    fn = cm[i, :].sum() - tp
    tn = cm.sum() - (tp + fp + fn)

    # Compute Specificity and Negative Predictive Value (NPV)
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    npv = tn / (tn + fn) if (tn + fn) > 0 else 0

    specificities.append(specificity)
    npvs.append(npv)

# Compute Macro Specificity and NPV
macro_specificity = np.mean(specificities)
macro_npv = np.mean(npvs)

# Print Standard Metrics
print("\n🟢 Model Used: Softmax Classifier (Multinomial Logistic Regression)")
print(f"\n✅ Accuracy: {accuracy:.4f}")
print(f"✅ Precision: {precision:.4f}")
print(f"✅ Recall (Sensitivity): {recall:.4f}")
print(f"✅ F1 Score: {f1:.4f}")

# Print Additional Metrics
print(f"\n🔹 Macro Negative Predictive Value (NPV): {macro_npv:.4f}")
print(f"🔹 Macro Specificity: {macro_specificity:.4f}")

# Print Detailed Classification Report
print("\n📊 Classification Report:\n", classification_report(y_test, y_pred, digits=4))

```
